# **Hirarical_Clustering**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# 1.Employee Data Entry (Technical Assessment, Communication Assessment) # Rating: A, B, C, D, E
X = np.array([
    [2, 4],  # A
    [2, 5],  # B
    [8, 9],  # C
    [9, 8],  # D
    [5, 6]   # E
])

labels = ['Employee A', 'Employee B', 'Employee C', 'Employee D', 'Employee E']
```

```python
# 2. Hirarical clustering
linked = linkage(X, method='median')

# 3. Ploting the tree of the model (Dendrogram)
plt.figure(figsize=(10, 6))
dendrogram(linked,
           orientation='top',
           labels=labels,
           distance_sort='descending',
           show_leaf_counts=True)
```

```python

# 4. Make the plot  more  clear
plt.title('Hierarchical Clustering Dendrogram (Median Linkage)', fontsize=14, weight='bold')
plt.xlabel('Employees', fontsize=12)
plt.ylabel('Merge Distance (Euclidean)', fontsize=12)

plt.axhline(y=4.5, color='r', linestyle='--', label='Cut Line (K=2)')
plt.legend()

# Show me thediagram
plt.tight_layout()
plt.show()
```
